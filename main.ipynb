```python
from datasets import Dataset, DatasetDict

train_dataset = Dataset.from_pandas(mer_d)
test_dataset = Dataset.from_pandas(mer1_d)

dataset_dict = DatasetDict({"train": train_dataset, "test": test_dataset})

# Print the structure
print(dataset_dict)
```

```python
train_dataset = {"question": X_t, "answers": y_t}
len(train_dataset["question"])
```

```python
def preprocess_function(examples):
    questions = [q.strip() for q in examples["Questions"]]
    answers = examples["Answers"]
    inputs = tokenizer(
        questions,
        answers,
        max_length=384,
        truncation="only_second",
        padding="max_length",
    )

    start_positions = []
    end_positions = []

    for i, question in enumerate(questions):
        question_len = len(tokenizer(question)['input_ids'])
        answer_len = len(tokenizer(answers[i])['input_ids'])  # Assuming each answer is a string

        # Assign start and end positions for answer spans within the question
        start_positions.append(question_len)
        end_positions.append(question_len + answer_len - 1)

    inputs["start_positions"] = start_positions
    inputs["end_positions"] = end_positions
    return inputs
```

```python
tokenized_squad = dataset_dict.map(preprocess_function, batched=True, remove_columns=dataset_dict["train"].column_names)
```

```python
tokenized_squad["train"]
```

```python
from transformers import DefaultDataCollator

data_collator = DefaultDataCollator()
```

```python
from transformers import AutoModelForQuestionAnswering, TrainingArguments, Trainer

model = AutoModelForQuestionAnswering.from_pretrained("distilbert/distilbert-base-uncased")
```

```python
training_args = TrainingArguments(
    output_dir="my_awesome_qa_model_reddit",
    evaluation_strategy="epoch",
    learning_rate=2e-5,
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    num_train_epochs=3,
    weight_decay=0.01,
    push_to_hub=True,
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_squad["train"],
    eval_dataset=tokenized_squad["test"],
    tokenizer=tokenizer,
    data_collator=data_collator,
)

trainer.train()
```

```python
trainer.push_to_hub()
```

```python
import torch
from transformers import AutoModelForQuestionAnswering

model = AutoModelForQuestionAnswering.from_pretrained("my_awesome_qa_model")
with torch.no_grad():
    outputs = model(**inputs)
```
