```python
import json
from neo4j import GraphDatabase

# Connect to Neo4j
uri = "bolt://localhost:7687"  # Update as necessary
username = "Graph"
password = "password"  # Update as necessary
driver = GraphDatabase.driver(uri, auth=(username, password))

def create_match(tx, match_data):
    tx.run("""
    CREATE (m:Match {id: $id, season: $season, match_type: $match_type, result: $result})
    """, match_data)

def create_teams(tx, teams, match_id):
    for team in teams:
        tx.run("""
        CREATE (t:Team {name: $team_name})
        MERGE (m:Match {id: $match_id})
        CREATE (m)-[:HAS_TEAM]->(t)
        """, team_name=team, match_id=match_id)

def create_players(tx, players, match_id):
    for team, player_list in players.items():
        for player in player_list:
            tx.run("""
            CREATE (p:Player {name: $player_name, team: $team_name})
            MERGE (t:Team {name: $team_name})
            CREATE (t)-[:HAS_PLAYER]->(p)
            """, player_name=player, team_name=team)

def create_overs(tx, overs, match_id):
    for over in overs:
        tx.run("""
        CREATE (o:Over {number: $over_number})
        MERGE (m:Match {id: $match_id})
        CREATE (m)-[:HAS_OVER]->(o)
        """, over_number=over['over'], match_id=match_id)
        create_deliveries(tx, over['deliveries'], match_id, over['over'])

def create_deliveries(tx, deliveries, match_id, over_number):
    for delivery in deliveries:
        tx.run("""
        CREATE (d:Delivery {batter: $batter, bowler: $bowler, non_striker: $non_striker, runs: $runs, extras: $extras, total: $total})
        MERGE (o:Over {number: $over_number})
        CREATE (o)-[:HAS_DELIVERY]->(d)
        """, batter=delivery['batter'], bowler=delivery['bowler'], non_striker=delivery['non_striker'],
             runs=delivery['runs']['batter'], extras=delivery['runs']['extras'], total=delivery['runs']['total'],
             over_number=over_number)

def main():
    # Load data from JSON file
    with open(r"C:\Users\Yashwanth\projects\LLM sports\recently_added_2_male_json\1415751.json", 'r') as f:
        data = json.load(f)

    match_data = {
        "season": data['info']['season'],
        "match_type": data['info']['match_type'],
    }

    teams = data['info']['teams']

    players = data['info']['players']

    overs = []
    for innings in data['innings']:
        for over in innings['overs']:
            overs.append({
                "over": over['over'],
                "deliveries": over['deliveries']
            })

    with driver.session() as session:
        session.write_transaction(create_match, match_data)
        session.write_transaction(create_teams, teams, match_data['id'])
        session.write_transaction(create_players, players, match_data['id'])
        session.write_transaction(create_overs, overs, match_data['id'])

if __name__ == "__main__":
    main()
```

```python
import re
from num2words import num2words
from PyPDF2 import PdfReader
from fpdf import FPDF

# Function to convert number to words
def number_to_words(match):
    num = int(match.group(0))
    return num2words(num)

# Read the PDF file
file_path = r"C:\Users\Yashwanth\Downloads\Formatted_Match_Info.pdf"
pdf_reader = PdfReader(open(file_path, 'rb'))

# Prepare the text content
content = ''
for page in pdf_reader.pages:
    content += page.extract_text() + '\n'

# Process the content
processed_content = re.sub(r'\b\d+\b', number_to_words, content)

# Save the processed content to a new PDF
class PDF(FPDF):
    def header(self):
        self.set_font('Arial', 'B', 12)
        self.cell(0, 10, 'Formatted Match Info', 0, 1, 'C')

    def chapter_title(self, title):
        self.set_font('Arial', 'B', 12)
        self.cell(0, 10, title, 0, 1, 'L')

    def chapter_body(self, body):
        self.set_font('Arial', '', 12)
        self.multi_cell(0, 10, body)
        self.ln()

# Create a PDF instance
pdf = PDF()
pdf.add_page()
pdf.chapter_body(processed_content)

# Save the new PDF
output_path = 'Formatted_Match_Info_Converted.pdf'
pdf.output(output_path)

print(f"Processed PDF saved to {output_path}")
```
