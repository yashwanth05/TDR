```python
import fitz  # PyMuPDF
import chromadb

def extract_text_from_pdf(pdf_path):
    doc = fitz.open(pdf_path)
    text = ""
    for page_num in range(doc.page_count):
        page = doc.load_page(page_num)
        text += page.get_text()
    return text

# Path to your PDF file
pdf_path = r"C:\Users\Yashwanth\Downloads\Afganisthan vs England.pdf"
extracted_text = extract_text_from_pdf(pdf_path)

# Split the extracted text into individual documents
documents = extracted_text.strip().split('\n')

# Generate unique IDs for each document
ids = [f"id{i+1}" for i in range(len(documents))]

# Initialize the ChromaDB client
chroma_client = chromadb.Client()

# Get or create a collection
collection = chroma_client.get_or_create_collection(name="cricket_match_collection")

# Upsert documents into the collection
collection.upsert(
    documents=documents,
    ids=ids
)

# Query the collection for a specific over and ball
query_text = "what happened in ball 1 over 0?"
results = collection.query(
    query_texts=[query_text], # Chroma will embed this for you
    n_results=1 # Number of results to return
)

# Print the results
print(results)
```
