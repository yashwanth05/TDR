```python
import os
import json

# Function to convert string values to integers and floats
def convert_values(entry):
    entry['R'] = int(entry['R'])
    entry['B'] = int(entry['B'])
    entry['4s'] = int(entry['4s'])
    entry['6s'] = int(entry['6s'])
    entry['SR'] = float(entry['SR'])

# Function to process all files in a folder
def process_folder(folder_path):
    for file_name in os.listdir(folder_path):
        if file_name.endswith('.json'):
            file_path = os.path.join(folder_path, file_name)
            with open(file_path, 'r') as file:
                data = json.load(file)
            
            for entry in data:
                convert_values(entry)
            
            with open(file_path, 'w') as file:
                json.dump(data, file, indent=2)

# Replace 'folder_path' with the path to your folder containing the JSON files
folder_path = r"E:\TDL_Dataset\GG"

process_folder(folder_path)
```
